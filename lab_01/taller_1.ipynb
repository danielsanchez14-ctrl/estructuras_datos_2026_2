# Creación de una matriz de 100.000x100.000 en disco

```python

"""
    Generación de Matriz 100,000×100,000 en Disco

    Crea una matriz con filas alternadas de 0s y 1s (10GB aprox.) usando un generador
    lazy que produce filas bajo demanda. writelines() consume el generador
    escribiendo eficientemente en disco sin cargar todo en memoria.
    La creación de la matriz se hace con caracteres "0" y "1" ASCII, cada uno de estos caracteres mide 1 Byte.

    Archivo resultante:
    - Línea 1: Metadatos (#FILAS=100000 COLUMNAS=100000)
    - Líneas 2+: Matriz (filas pares=0s, impares=1s)
    - Cada fila se separa de otra mediante un salto de línea
"""
    
def crear_matriz():
    FILAS = 100_000
    COLUMNAS = 100_000
    ARCHIVO = "matriz.txt"

    # Precomputamos las filas SIN espacios
    fila_par = ("0" * COLUMNAS) + "\n"
    fila_impar = ("1" * COLUMNAS) + "\n"

    with open(ARCHIVO, "w", encoding="utf-8", newline="") as f:
        f.write(f"#FILAS={FILAS} COLUMNAS={COLUMNAS}\n")

        # Generador que solo se ejecuta cuando se usa en writelines()
        filas = (fila_par if i % 2 == 0 else fila_impar for i in range(FILAS))

        # Escribe en disco la fila de la matriz, activando el generador un paso a la vez
        f.writelines(filas)
```

```python
## Ejecutamos la función de creación
crear_matriz()
```

# Funciones de manipulación y lectura de datos

Las siguientes funciones permiten leer la matriz, de manera tal que no se cargue toda en la memoria principal.

```python
"""
    Verifica que las dimensiones declaradas en la metadata coincidan
    con las dimensiones reales del archivo en disco.
"""
def verificar_matriz():
    ARCHIVO = "matriz.txt"
    import os
    with open(ARCHIVO, "rb") as f:
        header_len, filas, columnas, row_len = _info_archivo(f)
        tamaño_esperado = header_len + filas * row_len
        tamaño_real = os.fstat(f.fileno()).st_size #Tamaño del archivo en bytes según el sistema de archivos

        if tamaño_real != tamaño_esperado:
            print(f"ERROR: el archivo pesa {tamaño_real} bytes, se esperaban {tamaño_esperado}")
            return False

        print(f"OK: {filas} filas x {columnas} columnas confirmadas")
        return True
```

```python
# Validación
verificar_matriz()
```

```python
"""
    Lee el archivo y calcula el "mapa" necesario para acceder
    a cualquier fila o celda directamente con seek() (aritmética de offsets), sin tener que leer
    el archivo desde el principio.

    Devuelve una tupla con:
      - header_len: cuántos bytes ocupa la línea de metadata (dónde termina
        el header y empieza la matriz).
      - filas: cantidad de filas de la matriz.
      - columnas: cantidad de columnas de la matriz.
      - row_len: cuántos bytes ocupa cada fila en disco (columnas + 1 por
        el salto de línea "\n" que la separa de la siguiente).

    Asume que el archivo está abierto en modo binario ("rb" o "r+b") y que
    todo su contenido es ASCII (cada carácter ocupa exactamente 1 byte),
    condición necesaria para que los offsets calculados con estos valores
    sean correctos.
"""

def _info_archivo(f):
    f.seek(0) #Se ubica en el byte 0 del archivo (justo al principio)
    metadata = f.readline()              # bytes, incluye el "\n" (Columna con la metadata)
    header_len = len(metadata)           # largo exacto de la metadata en bytes
    texto = metadata.decode("utf-8").strip().replace("#", "") #Convierte los bytes a texto y quita espacios en blanco al principio y al final.
    partes = texto.split() # Separa ese texto por espacios, es una lista

    # Obtiene el número de filas y columnas
    filas = int(partes[0].split("=")[1])
    columnas = int(partes[1].split("=")[1])
    
    # Calcula cuánto ocupa cada fila en bytes
    row_len = columnas + 1             # +1 por el "\n" de cada fila
    return header_len, filas, columnas, row_len
```

```python
"""
    Obtiene los metadatos de la matriz generada, que corresponden a sus dimensiones (# Filas, # Columnas)
"""
def leer_metadata():
    ARCHIVO = "matriz.txt"
    with open(ARCHIVO, "r", encoding="utf-8") as f:

        # Leer la primera línea del archivo (aquí se encuentra la metadata)
        metadata = f.readline().strip()

        partes = metadata.replace("#", "").split() #Reemplaza el "#" por cadena vacía y separa en dos subcadenas (usa el espacio como separador)
        print(partes[0], partes[1]) #Imprimir la información en pantalla

```

```python
# Ejecución de la función
leer_metadata()
```

```python
"""
    Obtiene una fila de la matriz según su número de fila (iniciando en 1) y una cantidad específica de entradas de esa fila
"""
def mostrar_fila(numero_fila, cantidad_entradas=100):
    ARCHIVO = "matriz.txt"
    with open(ARCHIVO, "rb") as f:

        # Usar función auxiliar
        header_len, filas, columnas, row_len = _info_archivo(f)
        

        # Calcular el byte exacto donde empieza la fila pedida y saltar ahí
        offset = header_len + (numero_fila - 1) * row_len
        f.seek(offset)

        # Leer solo los "columnas" o bytes de esa fila (sin el "\n")
        fila = f.read(columnas).decode("utf-8")

        print(f"Fila: {numero_fila}")
        print(f"({len(fila)} elementos en total)")
        print(fila[:cantidad_entradas])
```

```python
# Ejecución de la función
mostrar_fila(100_000, cantidad_entradas=100_000)
```

```python
"""
    Muestra una porción de la matriz, tomando un intervalo de filas [i, j] y un intervalo de columnas [k, z]
    
    El conteo de filas y columnas inicia en 1
"""
def mostrar_bloque(fila_inicio, fila_fin, columna_inicio, columna_fin):
    ARCHIVO = "matriz.txt"
    with open(ARCHIVO, "rb") as f:
        # Toda la info del archivo sale de la función auxiliar
        header_len, filas, columnas, row_len = _info_archivo(f)

        # Cuántos caracteres hay que leer por fila (el ancho del bloque)
        ancho = columna_fin - columna_inicio + 1

        for numero_fila in range(fila_inicio, fila_fin + 1):
            # Byte exacto, inicio de la fila + desplazamiento hasta la columna pedida
            offset = header_len + (numero_fila - 1) * row_len + (columna_inicio - 1)
            f.seek(offset)
            bloque = f.read(ancho).decode("utf-8")
            print(bloque)
```

```python
mostrar_bloque(1, 100, 1, 100_000)
```

```python
"""
    Obtiene y muestra un elemento individual de la matriz, dada una posición de fila y una de columna.

    El conteo de filas y columnas inicia en 1.
"""
def mostrar_entrada(fila, columna):
    ARCHIVO = "matriz.txt"

    with open(ARCHIVO, "rb") as f:
        # Toda la info del archivo sale de la función auxiliar
        header_len, filas, columnas, row_len = _info_archivo(f)

        # Byte exacto, inicio de la fila + desplazamiento hasta la columna pedida
        offset = header_len + (fila - 1) * row_len + (columna - 1)
        f.seek(offset)

        # Leer un solo byte, el elemento puntual
        elemento = f.read(1).decode("utf-8")

        print(elemento)
```

```python
# Ejecución de la función
mostrar_entrada(1, 3)
```

```python
"""
    Modifica un elemento individual de la matriz, dada una posición de fila y una de columna,
    y el nuevo valor a escribir ("0" o "1").

    El conteo de filas y columnas inicia en 1.
"""
def modificar_elemento(fila, columna, valor):
    ARCHIVO = "matriz.txt"
    if valor not in ("0", "1"):
        print("Error: el valor debe ser 0 o 1.")
        return

    with open(ARCHIVO, "r+b") as f:
        # Toda la info del archivo sale de la función auxiliar
        header_len, filas, columnas, row_len = _info_archivo(f)

        # Byte exacto, inicio de la fila + desplazamiento hasta la columna pedida
        offset = header_len + (fila - 1) * row_len + (columna - 1)
        f.seek(offset)

        # Escribir un solo byte encima del valor anterior
        f.write(valor.encode("utf-8"))
```

```python
# Ejemplo, modifiquemos la entrada (fila 1, columna 3), debe observarse un "1" en medio de muchos ceros.
modificar_elemento(1, 3, "1")
mostrar_fila(1, 100)
```
